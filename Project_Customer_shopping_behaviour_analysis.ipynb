```python
#Which category has the highest purchase amount
category_count=data['Category'].value_counts()
plt.figure(figsize=(6,4))
plt.bar(category_count.index,category_count.values,color=["red","green","blue"])
plt.title("Highest purchase amount by category")
plt.xlabel("Category")
plt.ylabel("Count")
plt.show()
```

```python
#Category wise sales for each season
Season_sales=data.groupby(['Category','Season']).size().unstack()
Season_sales.plot(kind='bar',figsize=(7,4))
plt.title("Category wise sales for each season")
plt.xlabel("Category")
plt.ylabel("Season")
plt.show()
```

```python
#Gender Distribution
gender_count=data['Gender'].value_counts()
plt.figure(figsize=(6,6))
plt.pie(gender_count,labels=gender_count.index,autopct='%1.1f%%')
plt.title("Gender Distribution")
plt.show()
```

```python
#Age vs Purchase amount
Age_Purchase=data.groupby("Age")['Purchase Amount (USD)'].sum()
plt.scatter(Age_Purchase.index,Age_Purchase.values)
plt.title("Age vs Purchase Amount")
plt.xlabel("Age")
plt.ylabel("Purchase Amount")
plt.xticks(rotation=90)
plt.show()
```

```python
#Review rating distribution by category
sns.boxplot(x='Category',y='Review Rating',data=data)
plt.title('Review Rating distribution by category')
plt.xlabel('Category')
plt.ylabel('Review Rating')
plt.show()
```

```python
#Which payment method is used the most
Payment_method=data['Payment Method'].value_counts()
plt.figure(figsize=(6,6))
plt.pie(Payment_method,labels=Payment_method.index,autopct='%1.1f%%')
plt.title("Payment method distribution")
plt.show()
```

```python
#How does purchase amount vary across seasons
seasons_purchase=data["Season"].value_counts()
plt.figure(figsize=(6,4))
plt.bar(seasons_purchase.index,seasons_purchase.values,color=["red","green","blue"])
plt.title("Season wise Purchase distribution")
plt.xlabel("Season")
plt.ylabel("count")
plt.show()
```

```python
#Frequency of purchases over customer age
Age_Purchase=data.groupby("Frequency of Purchases")["Age"].mean()
plt.plot(Age_Purchase.index,Age_Purchase.values)
plt.title("Purchase Frequency over Customer Age")
plt.xlabel("Purchase Frequency")
plt.ylabel("Avg Age")
plt.xticks(rotation=90)
plt.show()
```
